````python
# 问题2 第三问 Python 实现归并排序，并输出运行结果

def merge_sort(A):
    if len(A) <= 1:
        return A
    
    mid = len(A) // 2
    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
            
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# 测试示例
A = [5, 2, 9, 1, 5, 6]
print("排序前:", A)
sorted_A = merge_sort(A)
print("排序后:", sorted_A)

# 输出算法时间复杂度
print("算法时间复杂度: O(n log n)")
````

profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:

````python
# 问题3 Python实现：计算最大利润并输出运行结果

def get_max_profit(profits):
    if not profits:
        return 0, None, None
        
    max_profit = float('-inf')  # 全局最大利润
    current_profit = 0           # 当前累加利润
    
    temp_buy_day = 0             # 当前区间的潜在买入日
    best_buy_day = 0             # 最优买入日
    best_sell_day = 0            # 最优卖出日
    
    for i in range(len(profits)):
        current_profit += profits[i]
        
        # 更新最大利润及对应的买卖日期
        if current_profit > max_profit:
            max_profit = current_profit
            best_buy_day = temp_buy_day
            best_sell_day = i
            
        # 如果当前累计利润小于0，重置区间
        if current_profit < 0:
            current_profit = 0
            temp_buy_day = i + 1
            
    # 日期加1转换成自然日编号（从1开始）
    return max_profit, best_buy_day + 1, best_sell_day + 1

# 测试示例
profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]
max_profit, buy_day, sell_day = get_max_profit(profits)

print("最大利润:", max_profit)
print("买入日:", buy_day)
print("卖出日:", sell_day)

# 输出算法时间复杂度
print("算法时间复杂度: O(n)")
````

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:

````python
# 1. O(log n) 找出一个局部最小值的思路：
# 因为已知 A[0]>A[1] 且 A[n-2]<A[n-1]，两端呈现向内下降趋势，中间必有谷底。
# 采用二分查找：检查 mid，若 A[mid] 比左右都小则是局部最小。
# 若 A[mid] > A[mid+1]，说明处于下降坡，右侧必有谷底，向右查找；反之向左查找。

def find_one_local_minimum(A):
    if not A or len(A) < 3:
        return None
    left, right = 1, len(A) - 2
    while left <= right:
        mid = (left + right) // 2
        if A[mid] < A[mid-1] and A[mid] < A[mid+1]:
            return A[mid]
        if A[mid] > A[mid+1]:
            left = mid + 1
        else:
            right = mid - 1
    return None

# 2. 找出所有局部最小值的思路：
# 找所有局部最小值最坏情况下可能存在 O(n) 个，因此必须用 O(n) 的时间复杂度进行一次线性扫描。
# 从索引 1 到 n-2 遍历数组，依次判断当前元素是否严格小于左右两侧相邻元素。

def find_all_local_minima(A):
    res = []
    for i in range(1, len(A) - 1):
        if A[i] < A[i-1] and A[i] < A[i+1]:
            res.append(A[i])
    return res

# e.g. 测试用例
A = [9, 3, 7, 2, 1, 4, 5]
print(f"局部最小元素为 {', '.join(map(str, find_all_local_minima(A)))}")
````

## 问题 5

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。
2. 建立一个包含这n个数的堆（heap）, 并且调用 k 次Extract-min 按顺序输出最小的k个数。使用往空堆中不断插入元素的方法建立堆, 分析这种方法建堆的时间复杂度, 并使用python编程实现
3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:

````python
import heapq

# 1. 排序后输出最小的 k 个数
# 排序算法选择：归并排序或快速排序。
# 时间复杂度分析：整个数组排序的时间复杂度为 O(n log n)，切片取出前 k 个数的时间为 O(k)。总时间复杂度为 O(n log n)。

def get_k_smallest_by_sort(arr, k):
    sorted_arr = sorted(arr)
    return sorted_arr[:k]


# 2. 逐个插入建最小堆，并提取 k 次
# 建堆时间复杂度分析：向空堆中逐个插入 n 个元素，第 i 次插入的时间复杂度最坏为 O(log i)。
# 总时间复杂度为 log(1) + log(2) + ... + log(n) = log(n!)。根据斯特林公式，其时间复杂度等同于 O(n log n)。
# 提取 k 次最小元素耗时 O(k log n)。因此总时间复杂度为 O((n+k) log n)。

def get_k_smallest_by_insert_heap(arr, k):
    min_heap = []
    # 往空堆中不断插入元素建堆
    for num in arr:
        heapq.heappush(min_heap, num)
        
    # 调用 k 次 Extract-min
    res = []
    for _ in range(k):
        if min_heap:
            res.append(heapq.heappop(min_heap))
    return res


# 3. 海量数据找前k小
# 算法设计思路：维护一个大小为 k 的最大堆 (Max-Heap)。
# 遍历海量数据，若堆中元素不足 k 个，直接入堆；若达到 k 个，则将当前元素与堆顶（当前 k 个数中的最大值）比较。
# 如果当前元素比堆顶小，说明当前堆顶不可能是前 k 小的数，弹出堆顶，插入新元素。
# 时间复杂度：遍历 n 个元素，每次操作最大堆最多耗时 O(log k)，总时间复杂度为 O(n log k)。空间复杂度 O(k) 完美契合内存限制。
# （注：Python的 heapq 默认是最小堆，存入元素的相反数即可模拟最大堆）

def get_k_smallest_massive_data(data_stream, k):
    max_heap = []
    for num in data_stream:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -num)
        else:
            if num < -max_heap[0]: # 如果新来的数比堆里最大的数要小
                heapq.heappop(max_heap)
                heapq.heappush(max_heap, -num)
                
    # 堆中存放的是前 k 小的相反数，将其恢复并排序返回
    res = [-x for x in max_heap]
    return sorted(res)


# --- e.g. 测试用例 ---
arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3
print(f"1. 排序法寻找前{k}个最小数据为: {get_k_smallest_by_sort(arr, k)}")
print(f"2. 插入建堆法寻找前{k}个最小数据为: {get_k_smallest_by_insert_heap(arr, k)}")
print(f"3. 内存受限法寻找前{k}个最小数据为: {get_k_smallest_massive_data(arr, k)}")
````

## 问题 6

> 星球碰撞问题：现有n个星球，在同一条直线上运行，如数组A所示，元素的绝对值表示星球的质量，负数表示星球自右向左运动，正数表示星球自左向右运动，当两个星球相撞的时候，质量小的会消失，大的保持不变，**质量相同的两个星球碰撞后自右向左运动的星球消失，自左向右的星球保持不变**，假设所有星球的速度大小相同。
>
> $ A=[23,-8, 9, -3, -7, 9, -23, 22] $

请设计一个算法模拟星球的运行情况，输出最终的星球存续情况（输出一个数组），分析算法设计思路，计算时间复杂度，并基于python编程实现。

e.g.
---
1.  输入： A=[-3,-6,2,8, 5,-8,9,-2,1]， 输出：[-3, -6, 2, 8, 9, 1]

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:

````python
# 1. 算法设计思路：
# 使用栈模拟星球碰撞：
# - 遍历数组，正数（向右飞）直接入栈。
# - 负数（向左飞）与栈顶正数星球碰撞：
#   * 栈顶质量小：栈顶消失，继续比较
#   * 栈顶质量相等：负向星球消失，栈顶正向保留（题目要求右向左消失）
#   * 栈顶质量大：负向星球消失
# - 负向星球经过碰撞后仍存活，直接入栈
# 最终返回栈中正向星球和未消失的负向星球（顺序不变）
#
# 2. 时间复杂度 O(n)：每个星球最多入栈一次、出栈一次
#    空间复杂度 O(n)：最坏情况下所有星球都存入栈

def simulate_planets(A):
    stack = []
    for planet in A:
        if planet > 0:
            # 向右飞，直接入栈
            stack.append(planet)
        else:
            # 向左飞，开始碰撞
            survived = True
            while stack and stack[-1] > 0:
                if stack[-1] < abs(planet):
                    # 栈顶质量小，弹出栈顶
                    stack.pop()
                elif stack[-1] == abs(planet):
                    # 栈顶质量等于负向星球，右向左消失，负向星球不入栈
                    survived = False
                    break
                else:
                    # 栈顶质量大，负向星球消失
                    survived = False
                    break
            if survived:
                stack.append(planet)
    return stack

# 测试
A1 = [-3, -6, 2, 8, 5, -8, 9, -2, 1]
A2 = [23, -8, 9, -3, -7, 9, -23, 22]

print(f"输入：{A1} 输出：{simulate_planets(A1)}")  # 预期: [-3, -6, 2, 9, 1]
print(f"输入：{A2} 输出：{simulate_planets(A2)}")  # 预期: [23, 22]
````

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

````python
# 算法设计思路：
# 使用动态规划(DP)求解。
# 维护一个一维数组 dp，其中 dp[w] 表示小推车载重量为 w 时所能装载的大理石最大价值。
# 遍历每一块大理石的重量和价值，从推车的最大承重开始逆序更新 dp 数组。
# 状态转移方程为：dp[w] = max(dp[w], dp[w - weight] + value)。
# 必须逆序遍历，以此保证在计算当前载重时，每一块大理石最多只被计算（放入）一次。

# add your code here
def get_max_value(marbles, max_capacity):
    dp = [0] * (max_capacity + 1)
    
    for weight, value in marbles:
        for w in range(max_capacity, weight - 1, -1):
            dp[w] = max(dp[w], dp[w - weight] + value)
            
    return dp[max_capacity]

# 测试用例
marbles = [(5, 10), (4, 40), (6, 30), (3, 50)]
max_capacity = 9
print(get_max_value(marbles, max_capacity))

# your algorithm time complexity is: O(n * W)，其中 n 为大理石的数量，W 为推车的最大承重
````

e.g.:

输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

2.

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

````python
# idea (算法设计思路):
# 正常二叉搜索树 (BST) 的中序遍历 (左-根-右) 结果必定是一个严格递增的序列。
# 如果有两个节点被错误交换，中序遍历中就会出现“降序”的逆序对（即 prev.val > curr.val）。
# 遍历过程中：
# 1. 如果这两个节点相邻，只会出现一次逆序对（如 1, 3, 2, 4 -> 3>2），我们记录 first=3, second=2。
# 2. 如果这两个节点不相邻，会出现两次逆序对（如 5, 2, 3, 4, 1 -> 5>2 和 4>1），第一次逆序记录 first=5，第二次逆序将 second 更新为 1。
# 遍历完成后，直接交换 first 和 second 节点的值即可恢复。

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# add your code here
def recoverTree(root):
    first = second = prev = None
    
    def inorder(node):
        nonlocal first, second, prev
        if not node:
            return
        
        inorder(node.left)
        
        # 寻找逆序对
        if prev and prev.val > node.val:
            if not first:
                first = prev  # 第一次遇到逆序，记录较大的节点
            second = node     # 记录较小的节点（如果遇到第二次逆序会更新它）
            
        prev = node
        
        inorder(node.right)
        
    inorder(root)
    
    # 交换两个错误节点的值
    if first and second:
        first.val, second.val = second.val, first.val

# your algorithm time complexity is: O(N) 
# 时间复杂度: O(N)，因为需要中序遍历每个节点一次（N为树中节点总数）。
# 空间复杂度: O(H)，H为树的高度，用于递归调用栈。最坏情况（退化为链表）为 O(N)，最好情况（平衡树）为 O(log N)。

def build_tree(values):
    if not values: return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        curr = queue.pop(0)
        if values[i] is not None:
            curr.left = TreeNode(values[i])
            queue.append(curr.left)
        i += 1
        if i < len(values) and values[i] is not None:
            curr.right = TreeNode(values[i])
            queue.append(curr.right)
        i += 1
    return root

def tree_to_list(root):
    if not root: return []
    res, queue = [], [root]
    while any(queue):
        curr = queue.pop(0)
        if curr:
            res.append(curr.val)
            queue.extend([curr.left, curr.right])
        else:
            res.append(None)
    # 移除末尾多余的 None
    while res and res[-1] is None:
        res.pop()
    return res

# 测试用例 1
root1 = build_tree([1, 3, None, None, 2])
recoverTree(root1)
print(f"输出 1: {tree_to_list(root1)}") 

# 测试用例 2
root2 = build_tree([3, 1, 4, None, None, 2])
recoverTree(root2)
print(f"输出 2: {tree_to_list(root2)}") 
````

````python
# 问题9：接雨水（此题目参考大模型）
# 算法：双指针法（从两端向中间逼近）
# 时间复杂度 O(n)，空间复杂度 O(1)

def trap(height):
    if not height:
        return 0
        
    left, right = 0, len(height) - 1
    left_max, right_max = 0, 0
    total_water = 0
    
    while left < right:
        if height[left] < height[right]:
            if height[left] >= left_max:
                left_max = height[left]
            else:
                total_water += left_max - height[left]
            left += 1
        else:
            if height[right] >= right_max:
                right_max = height[right]
            else:
                total_water += right_max - height[right]
            right -= 1
            
    return total_water

# --- 测试示例 ---
heights1 = [0,1,0,2,1,0,1,3,2,1,2,1]
heights2 = [4,2,0,3,2,5]

print("示例1：高度图", heights1, "接雨水:", trap(heights1))  # 输出: 6
print("示例2：高度图", heights2, "接雨水:", trap(heights2))  # 输出: 9

# 输出算法复杂度说明
print("算法时间复杂度: O(n)，空间复杂度: O(1)")
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

> 例如，从根节点到叶节点的路径 1 -> 2 -> 3 表示数字 123 。

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:

 输入：root = [1,2,3], 输出：25

解释：从根到叶子节点路径 1->2 代表数字 12，1->3 代表数字 13，因此和为 12 + 13 = 25 。

2.

 <div style="text-align:center;width:200px;"> <img alt="figure" src="./fig/hw3q10-2.jpg"> </div>

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:

````python
# 算法设计思路
# 采用深度优先搜索递归地遍历整棵树。
# 在遍历过程中，维护一个变量 curr_sum 来记录从根节点到当前节点路径所代表的数值。
# 每次向下走一层，新的数值计算方式为：curr_sum * 10 + 当前节点的值。
# 1. 如果当前节点是叶子节点（即没有子节点），直接返回当前的 curr_sum。
# 2. 如果当前节点不是叶子节点，则分别递归计算左子树和右子树生成的数字之和。

# add your code here
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def sumNumbers(root):
    def dfs(node, curr_sum):
        if not node:
            return 0
        
        # 计算当前路径代表的数
        curr_sum = curr_sum * 10 + node.val
        
        # 到达叶子节点，返回路径总值
        if not node.left and not node.right:
            return curr_sum
            
        # 递归求左右路径之和
        return dfs(node.left, curr_sum) + dfs(node.right, curr_sum)
    
    return dfs(root, 0)

# 测试 e.g. 1
root1 = TreeNode(1, TreeNode(2), TreeNode(3))
print(f"输出 1: {sumNumbers(root1)}")  # 期望 25

# 测试 e.g. 2
root2 = TreeNode(4, TreeNode(9, TreeNode(5), TreeNode(1)), TreeNode(0))
print(f"输出 2: {sumNumbers(root2)}")  # 期望 1026

# your algorithm time complexity is: O(n), 其中 n 为树中节点的总数，因为每个节点仅被访问一次。
````
